```python
import pandas as pd

sudan = pd.read_csv("evaluation/sudan_results_training.csv").drop_duplicates(subset="run_id", keep="last")
```

```python
run_names = sorted(sudan["run_name"].unique())

with open("models/completed_runs.txt", "w") as f:
    f.write("\n".join(run_names) + "\n")

print(f"Wrote {len(run_names)} completed run names to models/completed_runs.txt")
```
