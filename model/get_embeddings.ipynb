```python
import processing.acled_events_processing as acled
from processing.acled_text_processing import get_monthly_regional_embeddings, remove_dates, check_max_tokens
from transformers import AutoModel, AutoTokenizer, PreTrainedModel, PreTrainedTokenizer
```

```python
download = False
remove_abyei = False
k = 0.5
event_col = "event_type"
```

```python
_, _, raw_acled_df = acled.get_clean_data(
        download, remove_abyei, k, event_col
    )
```

```python
model_name = "eventdata-utd/ConfliBERT-scr-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
model.eval()  # Tells model I am not training

df_regex = remove_dates(raw_acled_df)
check_max_tokens(tokenizer, df_regex)
df_monthly = get_monthly_regional_embeddings(df_regex, tokenizer, model)
```

```python
df_monthly.to_pickle("../data/monthly_regional_embeddings.pkl")
```
